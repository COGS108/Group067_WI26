## Research Question

What combinations of academic credentials (undergraduate GPA and GMAT score) and professional experience (years of work experience and prior industry) are most strongly associated with MBA admission outcomes (Admit vs. Non-Admit) in the Wharton Class of 2025 dataset?



## Background and Prior Work


MBA admissions decisions are inherently multifactorial, combining academic performance, standardized test scores, and professional experience to evaluate an applicant’s readiness and potential. Top programs such as Wharton explicitly emphasize holistic review processes, where quantitative metrics are considered alongside work experience, leadership trajectory, and academic background. As a result, applicants are rarely evaluated based on a single credential; instead, admissions outcomes reflect how multiple factors interact within an applicant’s overall profile.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Prior research has consistently examined the role of academic metrics in MBA admissions and performance. A large meta-analysis by Kuncel et al. found that standardized tests such as the GMAT are moderately predictive of graduate academic success, particularly in quantitatively rigorous programs, while undergraduate GPA reflects sustained academic effort over time.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) However, the study also emphasized that test scores and GPA alone are insufficient to fully explain student outcomes, motivating broader evaluation frameworks that include professional experience and contextual factors.

Professional experience has been identified as another key component of MBA admissions decisions. Reports from the Graduate Management Admission Council (GMAC) indicate that years of work experience and industry background are used by admissions committees to assess leadership potential, maturity, and career progression.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) These reports suggest that applicants with fewer years of experience may still be competitive if they demonstrate strong academic preparation, while applicants with extensive work experience may offset weaker academic indicators. This supports the idea that combinations of credentials matter more than any single variable.

In applied data science contexts, several publicly available projects have analyzed MBA admissions datasets using exploratory data analysis and predictive modeling. For example, Kaggle-based projects examining MBA admissions data commonly use logistic regression and subgroup analysis to explore relationships between GPA, GMAT scores, work experience, and admission outcomes.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) These projects often find evidence of interaction effects, such as high GMAT scores compensating for lower GPA or work experience strengthening applications from non-traditional academic backgrounds. Building on this prior work, the present project focuses specifically on identifying which combinations of academic credentials and work experience are most strongly associated with admission in the Wharton Class of 2025 dataset.

While prior research and applied analyses provide useful insight into MBA admissions patterns, several limitations and threats to validity are worth noting. Many admissions-related datasets, including the Wharton Class of 2025 dataset used in this project, are synthetic or incomplete representations of real admissions processes, limiting causal interpretation of results.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Additionally, key qualitative factors such as essays, recommendation letters, and interviews—often emphasized by admissions committees—are not captured in structured datasets, potentially overstating the influence of measurable variables like GPA, GMAT scores, and work experience. Prior studies also caution that standardized test performance and academic metrics may reflect access to resources rather than underlying leadership or managerial potential, underscoring the importance of distinguishing evidence-based findings from admissions rhetoric when interpreting results.<a name="cite_ref-2b"></a>[<sup>2</sup>](#cite_note-2)

## Hypothesis


Hypothesis 1:
Applicants in the Wharton Class of 2025 dataset who demonstrate both strong academic credentials (signified by: higher undergraduate GPA and higher GMAT/GRE scores) and strong professional experience (signified by: more years of full-time work, higher position within companies, at stronger companies) will have the highest probability of admission. Applicants who are strong in only academics or only work experience will be less likely to be admitted than those who show strength in both areas.

Rationale:
Wharton is one of the most competitive MBA programs. Their applicant are likely among the most successful, so having both strong academic and professional experience will make for a stronger application. Strong academic performance shows preparation for the quantitative and analytical rigor of the program, while leadership experience shows future career potential.


Hypothesis 2:
Applicants with relatively weaker academic credentials (lower GPA or test scores) may still have competitive admission chances if they have exceptional professional experience, such as prestigious employers, entrepreneurial success, or significant leadership accomplishments.

Rationale:
It's a business school at the end of the day. Some people don't necessarily do well in school but are extremely successful people. Some people are also late bloomers, so they only recently started working very hard. Therefore, we believe lower academic credentials can sometimes be compensated for by strong professional experience.

## Data

Ideal dataset description

To answer questions about MBA admissions outcomes and potential patterns or disparities in admissions decisions, the ideal dataset would include detailed applicant-level information. Key variables would ideally include academic metrics (such as undergraduate GPA and standardized test scores like GMAT), demographic information (gender, race/ethnicity, and international status), professional background (years of work experience and prior industry), and the final admission decision (admit, waitlist, or deny). Additional useful variables could include undergraduate institution ranking, recommendation strength, essays scores, or interview performance to better capture non-academic factors. 
A sufficiently large dataset is needed to detect meaningful patterns across subgroups (at least 5,000–10,000 applicants) to ensure enough representation across majors, industries, and demographic categories. The data would be collected directly from MBA program application systems, where applicants self-report demographic information and academic history, and admissions committees record outcomes.
The dataset would be stored in a structured, tabular format (such as a CSV or relational database), where each row represents one applicant and each column represents a variable. This organization allows for efficient statistical analysis, visualization, and modeling using tools such as Python.

Chosen Real Dataset

We have already decided to use the MBA Admission Dataset for the Wharton Class of 2026, available on Kaggle: https://www.kaggle.com/datasets/taweilo/mba-admission-dataset/data 
This dataset is a synthetic dataset generated using summary statistics from Wharton’s Class of 2025 admissions data. It contains realistic applicant profiles while avoiding privacy concerns associated with real admissions records. The dataset is publicly available on Kaggle and can be downloaded as a ZIP file containing a CSV; no special permissions, applications, or approvals are required to use it.

Dataset Variables

 - application_id: Unique identifier for each application
 - gender: Applicant’s gender (Male, Female)
 - application_id: Unique identifier for each application
 - international: International student status (TRUE/FALSE)
 - gpa: Undergraduate GPA on a 4.0 scale
 - major: Undergraduate major (Business, STEM, Humanities)
 - race: Racial background (e.g., White, Black, Asian, Hispanic; null for international students)
 - gmat: GMAT score (out of 800)
 - work_exp: Number of years of work experience
 - work_industry: Industry of prior work experience (e.g., Consulting, Finance, Technology)
 - admission: Admission outcome (Admit, Waitlist; null indicates Deny)

 Observations

 The dataset contains 6,194 observations, with each row corresponding to a unique application ID. This sample size is large enough to explore trends in admissions outcomes across academic, professional, and demographic characteristics.

 Data Collection and Storage

 The data was found on Kaggle and downloaded as a CSV file. Since the dataset is synthetic, it does not involve direct data collection from applicants but instead simulates realistic admissions data based on published statistics. We will store and analyze the dataset locally by importing the CSV file into our analysis environment (e.g., VS Code using Python and pandas).
