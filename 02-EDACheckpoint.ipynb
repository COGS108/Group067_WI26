# Research Question

Which combination of browsing-related features (e.g., product page engagement, informational content depth, media value) and session context variables (e.g., month, weekend, special day proximity, traffic source, visitor type) is associated with the highest probability of generating revenue, and how do these optimal feature combinations vary across different shopping occasions?

## Background and Prior Work


Online shopping websites usually have a large number of visitors but only a small percentage of them actually make a purchase. That pattern is not surprising from a consumer perspective. Most people browse, compare options, and leave without buying. Because of this, researchers focus on identifying signals during a browsing session that indicate stronger purchase intent and understanding the conditions that increase the likelihood of conversion. Rather than treating a purchase as a single isolated event, many studies describe it as a process that includes searching, comparing alternatives, and evaluating information. For example, when buying a new pair of running shoes, a customer will likely open multiple tabs, compare prices and reviews, and revisit certain pages before deciding. This is why session-level features such as time spent on product pages, number of pages viewed, and types of content accessed are commonly used to predict purchases.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Research shows that browsing behavior within a single session can be highly informative for predicting whether a user will buy something. Studies published in *Decision Support Systems* demonstrate that combining browsing data with customer characteristics, such as whether a visitor is new or returning, improves prediction accuracy.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) Similarly, research in *Electronic Markets* finds that clickstream data, including navigation depth and page sequence patterns, can effectively predict purchase intention.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) These findings support the idea that engagement metrics alone are useful, but they are more powerful when analyzed alongside contextual variables.

More recent work suggests that the order and structure of browsing also matter. The way users move through a website can indicate whether they are casually exploring or shopping with a specific goal in mind. For example, directly searching for a product using the search bar may reflect clearer intent than navigating through the homepage and category pages. These browsing patterns can shift depending on promotions or specific shopping occasions, which implies that the most predictive engagement features may vary across contexts.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4)

Time-related factors also play a meaningful role in purchasing behavior. Reports from organizations such as the National Retail Federation and Adobe show that online spending increases substantially during holidays and major shopping events.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) <a name="cite_ref-6"></a>[<sup>6</sup>](#cite_note-6) This supports including variables such as month and proximity to special days when analyzing revenue outcomes. Retailers often track important dates, such as birthdays or seasonal events, and offer targeted promotions during those times to encourage purchases.

Traffic source and visitor type are also consistently linked to conversion differences. Direct traffic and returning visitors often have higher purchase rates, likely because they are already familiar with the brand or have stronger purchase intent. Google Analytics formally defines traffic channel categories (e.g., Direct, Organic Search, Referral), reinforcing the idea that acquisition source should be treated as a key predictor of conversion.<a name="cite_ref-7"></a>[<sup>7</sup>](#cite_note-7)

# Hypothesis



Revenue is most likely to happen when someone is genuinely interested in a product and the timing is right. If a user is spending real time on product pages, looking through details, and watching media, and on top of that they're a returning visitor, shopping on a weekend, close to a big sale, or coming from a source that shows they already have some intent, they're a lot more likely to actually buy something.

### Sub-Hypotheses

1. Engagement Effect: Users who actually look around and spend time on product pages should be more likely to convert than users who just bounce. The more someone engages, the more likely they are to buy.

2. Contextual Amplification: Engagement probably matters even more depending on when someone visits. Peak shopping months, weekends, and days around big promotions should push engaged users even closer to making a purchase.

3. Visitor Type Interaction: Returning visitors who are also highly engaged are probably the most likely to convert out of any group. They've been to the site before and came back, so pairing that with strong engagement should be a pretty reliable indicator of purchase intent.

4. Traffic Source Differentiation: The source someone comes from likely says a lot about their intent. Direct traffic probably signals stronger purchase intent than something like a random referral, so the same engagement behavior might lead to different conversion rates depending on where the user came from.

5. Occasion-Specific Variation: The features that best predict a purchase probably aren't the same across all time periods. What works during the holiday season likely looks pretty different from what drives conversions in the middle of summer, meaning the best predictors are probably context-dependent rather than universal.

**Data Description**

The dataset used in this project is the Online Shoppers Purchasing Intention Dataset obtained from the UCI Machine Learning Repository.
It contains 12,330 observations, where each row represents a single user session on an e-commerce website.
The dataset includes 18 variables that describe browsing behavior, session context, and whether the session resulted in a purchase (Revenue).
Because the data is session-level rather than user-level, each observation reflects one visit to the website rather than long-term customer behavior.

The most important metrics in this dataset relate to browsing engagement and purchase intent. Page visit variables such as Administrative, Informational, and ProductRelated measure the number of pages viewed in each category (unit: page counts).
Corresponding duration variables (e.g., ProductRelated_Duration) measure time spent on those pages in seconds.
BounceRates and ExitRates are proportions between 0 and 1 that represent the likelihood of leaving the website after viewing a page; values closer to 1 indicate higher likelihood of exiting.
PageValues represents the estimated monetary value of a page based on historical transaction data, with higher values suggesting stronger purchase intent.
SpecialDay ranges from 0 to 1 and represents how close the session occurred to a special shopping day (e.g., holidays), where values closer to 1 indicate proximity to that event.
The target variable, Revenue, is binary (1 = purchase occurred, 0 = no purchase).

### Data Relevance to the Research Question
This dataset is directly relevant to our research question because it contains both the independent variables and the outcome variable needed to examine purchasing behavior.
Our research question asks which combinations of browsing engagement features and session context variables are associated with the highest probability of generating revenue.
The dataset includes detailed engagement metrics such as ProductRelated page visits, time spent on pages, informational and administrative interactions, and PageValues, which together capture user interest and purchase intent.
In addition, the dataset includes contextual variables such as Month, Weekend, SpecialDay, TrafficType, and VisitorType. These variables allow us to analyze how purchasing behavior changes across different shopping occasions and traffic sources.
Most importantly, the dataset includes a binary Revenue variable indicating whether a purchase occurred. Because both predictors (engagement + context) and the outcome (purchase) are present in the same structured dataset, it enables direct analysis of how combinations of behavioral and contextual factors relate to conversion probability.
Although the data comes from a single e-commerce platform and is session-based rather than user-based, it provides sufficient behavioral granularity to meaningfully investigate patterns in online purchasing behavior. Therefore, it is well aligned with the goals of this project.

### Data Shortcomings
One potential concern with this dataset is that it originates from a single e-commerce platform, meaning results may not generalize to all online shopping environments.
Additionally, the dataset is session-based rather than customer-based, so repeat visitors may appear multiple times without being explicitly linked across sessions.
The dataset is also class imbalanced, with only about 15% of sessions resulting in a purchase, which may influence predictive modeling performance.
Finally, contextual variables such as income level, product type, and marketing campaigns are not included, limiting causal interpretation.
