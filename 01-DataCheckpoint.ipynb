### Wharton MBA Admissions Dataset (Synthetic – Modeled on Class of 2025 Summary Statistics) 

- **Dataset Name:** Wharton MBA Admissions Dataset (Synthetic – Modeled Class of 2025)
- **Link to the dataset:** https://www.kaggle.com/datasets/taweilo/mba-admission-dataset
- **Number of observations:** 6,194 applications
- **Number of variables:** 11 variables

- **Description of the variables most relevant to this project:**
  - **gpa:** Undergraduate GPA on a 4.0 scale. Values closer to 4.0 indicate stronger academic performance.
  - **gmat:** GMAT score (range 200–800). Higher scores indicate stronger standardized test performance.
  - **work_exp:** Years of full-time work experience.
  - **work_industry:** Industry category (Consulting, Finance, Technology, etc.).
  - **major:** Undergraduate major (Business, STEM, Humanities).
  - **international:** Boolean indicator of international applicant status.
  - **admission:** Admission outcome (Admit, Waitlist; null indicates Deny).

- **Descriptions of any shortcomings this dataset has with respect to the project:**
  - The dataset is synthetic and generated from published summary statistics rather than real admissions records.
  - It does not include qualitative factors such as essays, recommendation letters, or interview performance.
  - Some variables (such as race) contain missing values for international students.
  - Because the data is simulated, results should be interpreted as correlational rather than reflective of actual Wharton admissions decisions.


```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('data/00-raw/mba_admission.csv')

# Display first few rows
df.head()

# Show dataset size
print('Dataset shape:', df.shape)

# Check missing values
print('\nMissing values by column:')
print(df.isna().sum())

# Clean admission column (null means Deny)
df['admission'] = df['admission'].fillna('Deny')

# Check numeric ranges
print('\nSummary statistics:')
print(df[['gpa', 'gmat', 'work_exp']].describe())

# Flag potential outliers
print('\nInvalid GPA values:')
print(df[(df['gpa'] < 0) | (df['gpa'] > 4)])

print('\nInvalid GMAT values:')
print(df[(df['gmat'] < 200) | (df['gmat'] > 800)])

print('\nInvalid work experience values:')
print(df[df['work_exp'] < 0])

# Save cleaned dataset
df.to_csv('data/02-processed/mba_admission_clean.csv', index=False)
```

### Data Cleaning Decisions

Null values in the `admission` column represent denied applications according to the dataset documentation. These values were recoded as "Deny" to make admission outcomes explicit and ensure consistency for later analysis.

This transformation does not alter the distribution of outcomes but simply makes implicit denials explicit.
