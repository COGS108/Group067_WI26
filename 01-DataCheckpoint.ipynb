## Research Question

What combinations of academic credentials (undergraduate GPA and GMAT score) and professional experience (years of work experience and prior industry) are most strongly associated with MBA admission outcomes (Admit vs. Non-Admit) in the Wharton Class of 2025 dataset?



## Background and Prior Work


MBA admissions decisions are inherently multifactorial, combining academic performance, standardized test scores, and professional experience to evaluate an applicant’s readiness and potential. Top programs such as Wharton explicitly emphasize holistic review processes, where quantitative metrics are considered alongside work experience, leadership trajectory, and academic background. As a result, applicants are rarely evaluated based on a single credential; instead, admissions outcomes reflect how multiple factors interact within an applicant’s overall profile.<a name="cite_ref-1"></a>[<sup>1</sup>](#cite_note-1)

Prior research has consistently examined the role of academic metrics in MBA admissions and performance. A large meta-analysis by Kuncel et al. found that standardized tests such as the GMAT are moderately predictive of graduate academic success, particularly in quantitatively rigorous programs, while undergraduate GPA reflects sustained academic effort over time.<a name="cite_ref-2"></a>[<sup>2</sup>](#cite_note-2) However, the study also emphasized that test scores and GPA alone are insufficient to fully explain student outcomes, motivating broader evaluation frameworks that include professional experience and contextual factors.

Professional experience has been identified as another key component of MBA admissions decisions. Reports from the Graduate Management Admission Council (GMAC) indicate that years of work experience and industry background are used by admissions committees to assess leadership potential, maturity, and career progression.<a name="cite_ref-3"></a>[<sup>3</sup>](#cite_note-3) These reports suggest that applicants with fewer years of experience may still be competitive if they demonstrate strong academic preparation, while applicants with extensive work experience may offset weaker academic indicators. This supports the idea that combinations of credentials matter more than any single variable.

In applied data science contexts, several publicly available projects have analyzed MBA admissions datasets using exploratory data analysis and predictive modeling. For example, Kaggle-based projects examining MBA admissions data commonly use logistic regression and subgroup analysis to explore relationships between GPA, GMAT scores, work experience, and admission outcomes.<a name="cite_ref-4"></a>[<sup>4</sup>](#cite_note-4) These projects often find evidence of interaction effects, such as high GMAT scores compensating for lower GPA or work experience strengthening applications from non-traditional academic backgrounds. Building on this prior work, the present project focuses specifically on identifying which combinations of academic credentials and work experience are most strongly associated with admission in the Wharton Class of 2025 dataset.

While prior research and applied analyses provide useful insight into MBA admissions patterns, several limitations and threats to validity are worth noting. Many admissions-related datasets, including the Wharton Class of 2025 dataset used in this project, are synthetic or incomplete representations of real admissions processes, limiting causal interpretation of results.<a name="cite_ref-5"></a>[<sup>5</sup>](#cite_note-5) Additionally, key qualitative factors such as essays, recommendation letters, and interviews—often emphasized by admissions committees—are not captured in structured datasets, potentially overstating the influence of measurable variables like GPA, GMAT scores, and work experience. Prior studies also caution that standardized test performance and academic metrics may reflect access to resources rather than underlying leadership or managerial potential, underscoring the importance of distinguishing evidence-based findings from admissions rhetoric when interpreting results.<a name="cite_ref-2b"></a>[<sup>2</sup>](#cite_note-2)

### Wharton MBA Admissions Dataset (Synthetic – Modeled on Class of 2025 Summary Statistics) 

- **Dataset Name:** Wharton MBA Admissions Dataset (Synthetic – Modeled Class of 2025)
- **Link to the dataset:** https://www.kaggle.com/datasets/taweilo/mba-admission-dataset
- **Number of observations:** 6,194 applications
- **Number of variables:** 11 variables

- **Description of the variables most relevant to this project:**
  - **gpa:** Undergraduate GPA on a 4.0 scale. Values closer to 4.0 indicate stronger academic performance.
  - **gmat:** GMAT score (range 200–800). Higher scores indicate stronger standardized test performance.
  - **work_exp:** Years of full-time work experience.
  - **work_industry:** Industry category (Consulting, Finance, Technology, etc.).
  - **major:** Undergraduate major (Business, STEM, Humanities).
  - **international:** Boolean indicator of international applicant status.
  - **admission:** Admission outcome (Admit, Waitlist; null indicates Deny).

- **Descriptions of any shortcomings this dataset has with respect to the project:**
  - The dataset is synthetic and generated from published summary statistics rather than real admissions records.
  - It does not include qualitative factors such as essays, recommendation letters, or interview performance.
  - Some variables (such as race) contain missing values for international students.
  - Because the data is simulated, results should be interpreted as correlational rather than reflective of actual Wharton admissions decisions.


```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('data/00-raw/mba_admission.csv')

# Display first few rows
df.head()

# Show dataset size
print('Dataset shape:', df.shape)

# Check missing values
print('\nMissing values by column:')
print(df.isna().sum())

# Clean admission column (null means Deny)
df['admission'] = df['admission'].fillna('Deny')

# Check numeric ranges
print('\nSummary statistics:')
print(df[['gpa', 'gmat', 'work_exp']].describe())

# Flag potential outliers
print('\nInvalid GPA values:')
print(df[(df['gpa'] < 0) | (df['gpa'] > 4)])

print('\nInvalid GMAT values:')
print(df[(df['gmat'] < 200) | (df['gmat'] > 800)])

print('\nInvalid work experience values:')
print(df[df['work_exp'] < 0])

# Save cleaned dataset
df.to_csv('data/02-processed/mba_admission_clean.csv', index=False)
```

### Data Cleaning Decisions

Null values in the `admission` column represent denied applications according to the dataset documentation. These values were recoded as "Deny" to make admission outcomes explicit and ensure consistency for later analysis.

This transformation does not alter the distribution of outcomes but simply makes implicit denials explicit.
